## Contexto y problema de negocio

*¿Qué tarea se quería automatizar? ¿Quién usaba el resultado y qué decisión dependía de él?*

- ¿Cuál era el proceso antes (manual, reglas, otro sistema) y qué dolía: costo, tiempo, consistencia?

Crear un RAG local para apoyar al meson de ayuda que atendia a clientes. Requerimientos eran de usar codigo libre (open source) en maquinas propias, que soporte 4 idiomas (español, catalán, gallego y euskera). Montamos un RAG usando ollama y pgvector, con una UI y streamlit. Usamos una página de knowledge base para generar un dataset de Q&A y con este un dataset de embeddings.

- ¿Qué volumen de datos había y de qué tipo (textos cortos, documentos, conversaciones)?

Páginas de FAQ y knowledge base de la empresa, con preguntas y respuestas en 4 idiomas.

- ¿Cómo se definía el éxito para el negocio, y con qué métrica técnica se aproximaba?

Tiempos de respuesta y costes de herramienta. Evaluación qualitativa de trabajadoras de la mesa de ayuda, para este piloto


## Estrategia utilizada

*¿Cuál fue el enfoque, modelo o arquitectura?*

- Datos: ¿de dónde salieron las etiquetas o el corpus? ¿Cómo se trató su calidad?
- Enfoque: clasificador clásico, *embeddings*, *fine-tuning*, RAG, LLM con *prompting*... ¿Por qué ese y no otro? ¿Qué alternativas descartó?

Usamos un modelo con Ollama y pgvector, sin fine tuning, y prompting.

- Evaluación: ¿cómo se validó antes de producción? ¿Hubo revisión humana?

Si, pero cualitativo.

- Despliegue y monitoreo: ¿cómo se integró y cómo se vigilaba su desempeño?

Era un piloto y no logramos validar el producto economicamente.


## Resultados

- ¿Qué se logró, idealmente en números (precisión, horas ahorradas, cobertura)?

Era muy caro, se podía hacer con el file api de google o un modelo de openai, pero no era open source ni local, si no introduciamos datos de privados.

- ¿Qué no funcionó como se esperaba?
La respuesta de los modelos era lenta y no suficientemente buena, y el coste de la infraestructura era alto. No se pudo validar economicamente.
